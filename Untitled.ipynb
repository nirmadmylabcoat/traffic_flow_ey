```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file1 = "traffic_flow_data.csv"
file2 = "signal_timings_data.csv"

df_traffic = pd.read_csv(file1)
df_signals = pd.read_csv(file2)

df = pd.merge(df_traffic, df_signals, on="Intersection", how="left")

df.fillna(df.median(numeric_only=True), inplace=True)

df = df.sort_values(by="Hour")

plt.figure(figsize=(15, 6))

pivot_table = df.pivot_table(values="Vehicle_Count", index="Intersection", columns="Hour", aggfunc="sum")
pivot_table.fillna(pivot_table.min().min(), inplace=True)  

sns.heatmap(pivot_table, cmap="mako_r", annot=True, fmt=".0f", linewidths=1, linecolor="none")

plt.title("Traffic Heatmap (Hourly Vehicle Count)", fontsize=10, fontweight="bold", color="black")
plt.xlabel("Hour", fontsize=10)
plt.ylabel("Intersection", fontsize=10)
plt.xticks(rotation=45, fontsize=9)
plt.yticks(fontsize=9)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

traffic_file = "traffic_flow_data.csv"
signal_file = "signal_timings_data.csv"

df_traffic = pd.read_csv(traffic_file)
df_signals = pd.read_csv(signal_file)

vehicle_weights = {
    "Car": 1.0, 
    "Two-Wheeler": 0.5,
    "Company Bus": 2.5,  
    "Public Transport": 0.0,
    "Ride-Sharing": 1.2, 
}


df_traffic["Weighted_Count"] = df_traffic["Vehicle_Type"].map(vehicle_weights) * df_traffic["Vehicle_Count"]
df_traffic["Weighted_Count"].fillna(1, inplace=True)


df_traffic.fillna({"Weighted_Count": 57}, inplace=True) 

df_weighted = df_traffic.groupby(["Intersection", "Hour"]).agg({"Weighted_Count": "sum"}).reset_index()

df = pd.merge(df_weighted, df_signals, on="Intersection", how="left")
df.fillna(method="ffill", inplace=True)


max_weighted_count = df["Weighted_Count"].max()

def calculate_signal_duration(count, max_count, base_green=20, base_red=20):
    proportion = count / max_count if max_count > 0 else 0
    green_time = base_green + (proportion * 60) 
    red_time = base_red + ((1 - proportion) * 60)  
    return green_time, red_time

df[["Green_Light_Optimized", "Red_Light_Optimized"]] = df["Weighted_Count"].apply(
    lambda x: pd.Series(calculate_signal_duration(x, max_weighted_count))
)

df.sort_values(by=["Intersection", "Hour"], inplace=True)


fig, axes = plt.subplots(2, 1, figsize=(16, 10), sharex=True)
bar_width = 0.9


sns.barplot(x="Hour", y="Green_Light_Duration", hue="Intersection", data=df, ax=axes[0],
            palette="Blues", alpha=0.6, width=bar_width)
sns.barplot(x="Hour", y="Green_Light_Optimized", hue="Intersection", data=df, ax=axes[0],
            palette="mako", alpha=0.9, width=bar_width)
axes[0].set_title("Comparison of Green Light Duration (Previous vs Optimized)", fontsize=14, fontweight="bold", color="black")
axes[0].set_ylabel("Green Light Duration (s)")
axes[0].set_xlabel("Hours")
axes[0].set_xticks(range(0, 24, 1))
axes[0].set_xticklabels(range(0, 24, 1), rotation=45)
axes[0].legend(title="Intersection", bbox_to_anchor=(1.05, 1), loc="upper left")

sns.barplot(x="Hour", y="Red_Light_Duration", hue="Intersection", data=df, ax=axes[1],
            palette="Reds", alpha=0.6, width=bar_width)
sns.barplot(x="Hour", y="Red_Light_Optimized", hue="Intersection", data=df, ax=axes[1],
            palette="rocket", alpha=0.9, width=bar_width)
axes[1].set_title("Comparison of Red Light Duration (Previous vs Optimized)", fontsize=14, fontweight="bold", color="black")
axes[1].set_ylabel("Red Light Duration (s)")
axes[1].set_xlabel("Hours")
axes[1].set_xticks(range(0, 24, 1))  
axes[1].set_xticklabels(range(0, 24, 1), rotation=45)
axes[1].legend(title="Intersection", bbox_to_anchor=(1.05, 1), loc="upper left")

plt.tight_layout()
plt.show()
```
